## Variables

```python
# You don't need to declare a variable type.
i = 5
```

```python
# The type is inferred from the value assigned to the variable.
type(i)
```

```python
# You can use type hints to let others know the type of a variable.
i: int = 5
```

```python
# Here's a float.
f = 5.0
```

```python
# Type.
type(f)
```

```python
# Here's a string.
s = "Hello, world!"
```

```python
# Type.
type(s)
```

```python
# Complex numbers.
c = 1 + 2j
```

```python
# Square it.
c**2
```

```python
# What type will x have?
x = 5 + 5.0
```

```python
type(x)
```

## Strings

```python
# Strings can be declared with single or double quotes.
s1 = 'abc'
s2 = "abc"
```

```python
# Comparison.
s1 == s2
```

```python
# Concatenation.
s2 = "def"

s1 + s2
```

```python
# Slicing.
s = 'abcdefghijklmnopqrstuvwxyz'

# First five characters.
s[0:5]  
```

```python
# Every second character.
s[::2]
```

```python
# Formatted strings.
name = "Alice"

f"Hello, {name}!"
```

## Lists

```python
# List.
l = [1, 2, 3, 4, 5]

# Zero indexing.
l[2]
```

```python
# Slicing.
l[1:4]
```

```python
# Not quite a list.
a = range(0, 10)

a
```

```python
# But can be a list.
list(a)
```

## Conditionals

```python
from datetime import date

if date.today().weekday() == 1:
    print("Yes, it's Tuesday!")
elif date.today().weekday() == 0:
    print("Not today, but tomorrow is Tuesday.")
else:
    print("Nope, not Tuesday.")
```

## Loops

```python
for i in range(10):
    print(i, i**2, i**3)
```

## Functions

```python
# Define function.
def f(x):
    return x**2
```

```python
# Call function.
f(2)
```

```python
# Refer to function.
f
```

```python
# Functions are first class objects.
def twice(f, x):
    return f(f(x))
```

```python
twice(f, 2)
```

```python
# Closures.
def make_adder(n):
    def adder(x):
        return x + n
    return adder
```

```python
# Get a function back.
add5 = make_adder(5)
```

```python
# Call function.
add5(3)
```
