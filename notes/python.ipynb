## Strings

```python
# Strings can be declared with single or double quotes.
s1 = 'abc'
s2 = "abc"
```

```python
# Comparison.
s1 == s2
```

```python
# Concatenation.
s2 = "def"

s1 + s2
```

```python
# Every second character.
s[::2]
```

## Conditionals

[`if` statements on python.org](https://docs.python.org/3.14/tutorial/datastructures.html#more-on-lists)

```python
from datetime import date

if date.today().weekday() == 1:
    print("Yes, it's Tuesday!")
elif date.today().weekday() == 0:
    print("Not today, but tomorrow is Tuesday.")
else:
    print("Nope, not Tuesday.")
```

## Functions

[Defining Functions on python.org](https://docs.python.org/3.14/tutorial/controlflow.html#defining-functions)

```python
# Define function.
def f(x):
    return x**2
```

```python
# Call function.
f(2)
```

```python
# Refer to function.
f
```

```python
# Functions are first class objects.
def twice(fn, i):
    return fn(fn(i))
```

```python
twice(f, 2)
```

```python
# Closures.
def make_adder(n):
    def adder(x):
        return x + n
    return adder
```

```python
# Get a function back.
add5 = make_adder(5)
```

```python
# Call function.
add5(3)
```
