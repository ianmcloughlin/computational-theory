## Floating Point Numbers

![32-bit float](https://upload.wikimedia.org/wikipedia/commons/d/d2/Float_example.svg)

[Floating-Point Numbers, Real Python](https://realpython.com/python-bitwise-operators/#floating-point-numbers)  

[IEEE 754, WikiPedia](https://en.wikipedia.org/wiki/IEEE_754)  

[Floating-Point Arithmetic: Issues and Limitations](https://docs.python.org/3/tutorial/floatingpoint.html#floating-point-arithmetic-issues-and-limitations)

A float's bits split into three fields — sign $s$, exponent $e$, fraction $f$ — and the value they encode is:

$$(-1)^{s} \times 2^{(e - \text{bias})} \times (1.f)$$

A 32-bit float uses 1/8/23 bits (bias $127$); a 64-bit double uses 1/11/52 bits (bias $1023$). We decode a double into these fields later in the notebook.

````python
# Watch out for comparisons.
u = 0.3
v = 3.0 * 0.1
u - v == 0.0
````

````python
# Is close.
math.isclose(u, v)
````

## Bitwise Operators

Bitwise operators act independently on each bit of a number's binary representation.  
They are commonly used for masks, flags, and low-level manipulation, and are the core operations used in most hash functions.

https://wiki.python.org/moin/BitwiseOperators

````python
# Decimal 12.
u = 0b1100

# Decimal, binary, hex.
print(u, bin(u), hex(u))
````

````python
# Decimal 10.
v = 0b1010

# Decimal, binary, hex.
print(v, bin(v), hex(v))
````

````python
# Decimal 100.
w = 100

# Decimal, binary, hex.
print(w, bin(w), hex(w))
````

## Bitwise AND (`&`)


https://docs.python.org/3/library/stdtypes.html#bitwise-and  

Performs a bitwise AND on two numbers.  
Each bit is compared; if both are `1`, the result is `1`. Otherwise, it is `0`.

````python
# Decimal 12.
u = 0b1100

# Decimal 10.
v = 0b1010

# Decimal, binary, hex.
print(f"u: {u:04}   {u:08b} {u:02x}")

# Decimal, binary, hex.
print(f"v: {v:04}   {v:08b} {v:02x}")
````

````python
# 0b0001
result = u & v

# Decimal, binary, hex.
print(f"result: {result:04}   {result:08b} {result:02x}")
````

````python
# Print a using f-strings.
print(f'     u: {u:08b}')

# Print b using f-strings.
print(f'     v: {v:08b}')

# Print a line.
print('-' * 17)

# Print (u & v) using f-strings.
print(f' u & v: {(u & v):08b}')
````

## Bitwise OR (`|`)

Performs a bitwise OR.  
Each bit is compared; if either is `1`, the result is `1`.

https://docs.python.org/3/library/stdtypes.html#bitwise-or

````python
# Print a using f-strings.
print(f'     u: {u:08b}')

# Print b using f-strings.
print(f'     v: {v:08b}')

# Print a line.
print('-' * 17)

# Print (u | v) using f-strings.
print(f' u | v: {(u | v):08b}')
````

## Bitwise XOR (`^`)

Performs a bitwise XOR.  

Each bit is compared; if one is `1` and the other is `0`, the result is `1`.

https://docs.python.org/3/library/stdtypes.html#bitwise-xor

````python
# Print a using f-strings.
print(f'     u: {u:08b}')

# Print b using f-strings.
print(f'     v: {v:08b}')

# Print a line.
print('-' * 17)

# Print (u ^ v) using f-strings.
print(f' u ^ v: {(u ^ v):08b}')
````

## Bitwise NOT (`~`)

Performs a bitwise NOT.  

Inverts all bits: `0` becomes `1` and `1` becomes `0`.  

https://docs.python.org/3/library/stdtypes.html#bitwise-invert


````python
# Print a using f-strings.
print(f'     u: {u:08b}')

# Print a line.
print('-' * 17)

# Print ~u using f-strings.
print(f'    ~u: {~u:08b}')
````

## Negative Integers in Bits

Negative integers are represented in **two's complement**.  
To get the two's complement:
1. Invert all bits.
2. Add `1` to the result.

In two's complement:
- The leftmost bit indicates the sign.
- `0` means positive.
- `1` means negative.  

Use a bit mask like `0xf` to fix signed integer issues.  

In an $n$-bit two's complement representation, a negative number $x$ is stored as:

$$2^{n} + x$$

For example, with $n = 4$ bits, $-1$ is stored as $2^4 - 1 = 15 =$ `1111`.

````python
# Print binary of number 0 to 15 with the binary of the negative of the number.
for i in range(16):
    print(f'{i:02} {i:04b} {~i:06b} {(~i & 0xf):04b}')
````

## Left Shift (`<<`)
Shifts the bits to the left by a specified number of positions.  
Adds zeros to the right.

https://docs.python.org/3/library/stdtypes.html#bitwise-left-shift


````python
u = 0b010110111

# Print u using f-strings.
print(f'        u: {u:016b}')

# Print line.
print('-' * 28)

# Print u << 3 using f-strings.
print(f' (u << 3): {u << 3:016b}')
````

## Right Shift (`>>`)

Shifts the bits to the right by a specified number of positions.  
Drops bits from the right.  

https://docs.python.org/3/library/stdtypes.html#bitwise-right-shift


````python
u = 0b010110111

# Print u using f-strings.
print(f'        u: {u:016b}')

# Print line.
print('-' * 28)

# Print u >> 3 using f-strings.
print(f' (u >> 3): {u >> 3:016b}')
````

Python's `struct` module packs values into raw bytes (and unpacks them back), matching the memory layout used by C structs like the ones above. This lets us inspect the exact bits of a Python value.

````python
# A 64 bit integer.
large = 0xfedcba9876543210
````

````python
# Print large using f-strings.
print(f'large: {large:064b}')
````

````python
# Bit length.
large.bit_length()
````

## Byte Order

https://docs.python.org/3/library/struct.html#byte-order-size-and-alignment

Endianness is the order bytes are stored in memory: **little-endian** stores the least significant byte first, **big-endian** stores the most significant byte first. Networks conventionally use big-endian, also called network byte order.

````python
# Pack large into a struct with a 64-bit int, little-endian.
s = struct.pack('<Q', large)
````

````python
# Unpack in big endian.
large_big = struct.unpack('>Q', s)[0]
````

````python
# Print large using f-strings.
print(f's: {large:064b}')

# Print large_big using f-strings.
print(f't: {large_big:064b}')
````

````python
# Print s using f-strings.
print(f's: {hex(large)}')

# Print t using f-strings.
print(f't: {hex(large_big)}')
````

## Type Bit-Style Conversions

Floats in Python are usually 64-bit, which are typically called doubles:  
https://docs.python.org/3/tutorial/floatingpoint.html#representation-error  

How can we see the bits?  

![Double bits](https://upload.wikimedia.org/wikipedia/commons/a/a9/IEEE_754_Double_Floating_Point_Format.svg)

````python
# A float.
f = 23.0
````

````python
# Pack float into a struct.
s = struct.pack('>d', f)
````

````python
# Unpack struct into a float.
i = struct.unpack('>Q', s)[0]

# Show.
print(f"{f} {i:016x} {i:064b}")
````

````python
# Print f in exponential notation.
print(f'{f:e}')
````

````python
# Bit string.
fstr = f"{i:064b}"
````

````python
# Fraction.
sign, exp, frac = fstr[0], fstr[1:12], fstr[12:]

# Show.
print(sign, exp, frac)
````

````python
# To ints.
isign, iexp, ifrac = int(sign), int(exp, 2), int(frac, 2)

# Show.
print(isign, iexp, ifrac)
````

Reconstruct the original float from its sign, exponent, and fraction bits:

$$\text{value} = (-1)^{s} \times 2^{(e - 1023)} \times \left(1 + \frac{f}{2^{52}}\right)$$

````python
# Convert to float.
f_conv = ((-1) ** isign) * (2 ** (iexp - 1023)) * (1 + ifrac / 2 ** 52)
````

````python
# Show.
f_conv
````

## Summary

- Python `int`s are arbitrary-precision binary; `float`s are fixed-width IEEE 754.
- Bitwise operators (`& | ^ ~ << >>`) act bit-by-bit and underpin masks, flags, and hash functions.
- Negative integers use two's complement: $x$ is stored as $2^n + x$ in $n$ bits.
- `struct` converts between Python values and raw bytes, exposing byte order (endianness).
